## Task 2 — Cloud Resource Deployment Planning Agent
Goal: plan the sequence of cloud operations (resource allocation → VM provisioning → dependency install → model deployment → verification) that gets an AI application **deployed successfully**, minimizing total deployment time.

Two provisioning routes are modeled after *Resources Allocated* — building a fresh VM vs. using a prebuilt image (which then needs more dependency setup) — so the state space genuinely branches and the search has to pick the cheaper route.

```python
from task2_cloud_deployment import build_problem as build_deployment_problem

deployment_problem = build_deployment_problem()
print(describe_state_space(deployment_problem))
```

```python
deployment_result = uniform_cost_search(deployment_problem)
print(describe_solution(deployment_result, unit='min'))
```

## Task 3 — University Course Registration Planning Agent
Goal: guide a student through the mandatory academic workflow (authentication → prerequisite verification → course selection → fee verification → enrollment) so the student ends up **successfully enrolled**, minimizing registration steps.

Prerequisite verification can happen via an automated eligibility check or a (more expensive) manual advisor review, giving the planner a real choice.

```python
from task3_course_registration import build_problem as build_registration_problem

registration_problem = build_registration_problem()
print(describe_state_space(registration_problem))
```

```python
registration_result = uniform_cost_search(registration_problem)
print(describe_solution(registration_result, unit='step(s)'))
```
