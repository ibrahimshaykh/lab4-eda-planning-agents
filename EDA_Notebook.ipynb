# Exploratory Data Analysis — Titanic Dataset
Week 4 Lab Task 1: prototype notebook backing the Streamlit EDA Explorer app (`app.py`).

This notebook walks through the same steps as the deployed app: dataset ingestion, metadata inspection, attribute typing, and visualization.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('Titanic-Dataset.csv')
df.head()
```

## 1. Dataset Overview

```python
print('Shape (rows, cols):', df.shape)
df.dtypes
```

```python
missing = df.isna().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct}).sort_values('missing_count', ascending=False)
```

```python
numeric_df = df.select_dtypes(include='number')
summary = numeric_df.describe().T
summary['median'] = numeric_df.median()
summary[['mean', 'median', 'min', 'max']]
```

## 2. Attribute Typing

```python
def classify_column(df, column):
    return 'numerical' if pd.api.types.is_numeric_dtype(df[column]) else 'categorical'

{col: classify_column(df, col) for col in df.columns}
```

## 3. Visualization — Numerical Attribute (Age)

```python
fig = px.histogram(df, x='Age', nbins=30, marginal='box', template='plotly_white',
                    title="Distribution of 'Age'")
fig.update_layout(xaxis_title='Age', yaxis_title='Frequency')
fig.show()
```

## 4. Visualization — Categorical Attribute (Embarked)

```python
counts = df['Embarked'].value_counts(dropna=False).reset_index()
counts.columns = ['Embarked', 'count']
counts['percent'] = (counts['count'] / counts['count'].sum() * 100).round(2)

fig = px.bar(counts, x='Embarked', y='count', text='percent', template='plotly_white',
             title="Frequency of Categories in 'Embarked'")
fig.update_traces(texttemplate='%{text}%', textposition='outside')
fig.update_layout(xaxis_title='Embarked', yaxis_title='Count')
fig.show()
```
